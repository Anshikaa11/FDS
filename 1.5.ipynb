```python
import pandas as pd
from scipy.stats import t as t_dist
from scipy.stats import chi2 as chi2_dist

# Load your CSV file
df = pd.read_csv("dataset.csv")

# ====================================================
# T-TEST: Gender and Exam_Score
# H0: Mean Exam_Score for Male = Mean Exam_Score for Female
# H1: Mean Exam_Score for Male != Mean Exam_Score for Female
# ====================================================

male_group = df[df["Gender"] == "Male"]["Exam_Score"]
female_group = df[df["Gender"] == "Female"]["Exam_Score"]


def mean(data):
    total = 0

    for value in data:
        total += value

    return total / len(data)


def vari(data):
    mean_value = mean(data)
    total = 0

    for value in data:
        total += (value - mean_value) ** 2

    return total / (len(data) - 1)


mean_male = mean(male_group)
mean_female = mean(female_group)

var_male = vari(male_group)
var_female = vari(female_group)

n1 = len(male_group)
n2 = len(female_group)
t_value = (mean_male - mean_female) / (
    (var_male / n1 + var_female / n2) ** 0.5
)

print("----- T-TEST RESULTS -----")
print("Mean Exam Score (Male):", mean_male)
print("Mean Exam Score (Female):", mean_female)
print("Variance (Male):", var_male)
print("Variance (Female):", var_female)
print("n1:", n1)
print("n2:", n2)
print("Calculated t-value:", t_value)

# Welch degrees of freedom
df_num = (var_male / n1 + var_female / n2) ** 2

df_den = (
    ((var_male / n1) ** 2 / (n1 - 1))
    + ((var_female / n2) ** 2 / (n2 - 1))
)

degrees_freedom = df_num / df_den

print("Degrees of Freedom:", degrees_freedom)

alpha = 0.05
critical_t = t_dist.ppf(1 - alpha / 2, degrees_freedom)

print("Critical t-value:", critical_t)

if abs(t_value) > critical_t:
    print("Reject the Null Hypothesis")
    print("There is a significant difference in mean exam scores.")

else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant difference in mean exam scores.")


# ====================================================
# CHI-SQUARED TEST: Gender and Internet_Access
# H0: Gender and Internet_Access are independent
# H1: Gender and Internet_Access are associated
# ====================================================

male_yes = len(
    df[(df["Gender"] == "Male") &
       (df["Internet_Access"] == "Yes")]
)

male_no = len(
    df[(df["Gender"] == "Male") &
       (df["Internet_Access"] == "No")]
)

female_yes = len(
    df[(df["Gender"] == "Female") &
       (df["Internet_Access"] == "Yes")]
)

female_no = len(
    df[(df["Gender"] == "Female") &
       (df["Internet_Access"] == "No")]
)

# Observed frequency table
#             Internet Yes    Internet No
# Male           male_yes       male_no
# Female         female_yes     female_no

row1_total = male_yes + male_no
row2_total = female_yes + female_no

col1_total = male_yes + female_yes
col2_total = male_no + female_no

grand_total = row1_total + row2_total

# Expected frequencies
expected_male_yes = row1_total * col1_total / grand_total
expected_male_no = row1_total * col2_total / grand_total

expected_female_yes = row2_total * col1_total / grand_total
expected_female_no = row2_total * col2_total / grand_total

# Chi-square calculation
chi_square = (
    (male_yes - expected_male_yes) ** 2 / expected_male_yes
    + (male_no - expected_male_no) ** 2 / expected_male_no
    + (female_yes - expected_female_yes) ** 2 / expected_female_yes
    + (female_no - expected_female_no) ** 2 / expected_female_no
)

chi_df = 1
critical_chi = chi2_dist.ppf(1 - alpha, chi_df)

print("\n----- CHI-SQUARED TEST RESULTS -----")
print("Observed values:")
print("Male, Internet Yes:", male_yes)
print("Male, Internet No:", male_no)
print("Female, Internet Yes:", female_yes)
print("Female, Internet No:", female_no)

print("Chi-square value:", chi_square)
print("Degrees of Freedom:", chi_df)
print("Critical Chi-square value:", critical_chi)

if chi_square > critical_chi:
    print("Reject the Null Hypothesis")
    print("Gender and Internet Access are significantly associated.")

else:
    print("Fail to Reject the Null Hypothesis")
    print("Gender and Internet Access are not significantly associated.")
```
