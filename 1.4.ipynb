```python
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("dataset.csv")


# ==================== PMF ====================
# PMF gives probability of each discrete value

sleep = df["Sleep_Hours"]

pmf = sleep.value_counts() / len(sleep)

print("PMF of Sleep Hours:")
print(pmf)

plt.bar(pmf.index, pmf.values)

plt.title("PMF of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Probability")

plt.show()


# ==================== PDF ====================
# PDF shows the probability density of continuous data

score = df["Exam_Score"].dropna()

plt.hist(score, bins=10, density=True, edgecolor="black")

plt.title("PDF of Exam Score")
plt.xlabel("Exam Score")
plt.ylabel("Probability Density")

plt.show()


# ==================== CDF ====================
# CDF gives probability of getting a value
# less than or equal to a particular value

cdf = score.value_counts().sort_index().cumsum() / len(score)

print("\nCDF of Exam Score:")
print(cdf)

plt.plot(cdf.index, cdf.values)

plt.title("CDF of Exam Score")
plt.xlabel("Exam Score")
plt.ylabel("Cumulative Probability")

plt.show()
```
