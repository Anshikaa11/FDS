```python
import pandas as pd

# -------------------- READ DATASET --------------------

df = pd.read_csv("dataset.csv")

print("ORIGINAL DATASET")
display(df)

print("\nShape of Dataset:", df.shape)

print("\nColumn Names:")
print(list(df.columns))

print("\nData Types:")
print(df.dtypes)
```

```python
# ==========================================================
# 1. HANDLING MISSING VALUES
# ==========================================================

print("\n--------------- MISSING VALUES ---------------")

print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Distance_from_Home is categorical, so replace
# missing values with the mode

distance_mode = df["Distance_from_Home"].mode()[0]

df["Distance_from_Home"] = df["Distance_from_Home"].fillna(distance_mode)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# ==========================================================
# 2. HANDLING DUPLICATES
# ==========================================================

print("\n--------------- DUPLICATES ---------------")

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows =", duplicate_count)

# Remove duplicate rows

df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)
```

```python
# ==========================================================
# 3. HANDLING INCONSISTENT DATA
# ==========================================================

print("\n--------------- INCONSISTENT DATA ---------------")

categorical_columns = [
    "Parental_Involvement",
    "Access_to_Resources",
    "Extracurricular_Activities",
    "Motivation_Level",
    "Internet_Access",
    "School_Type",
    "Peer_Influence",
    "Learning_Disabilities",
    "Distance_from_Home",
    "Gender"
]

print("Unique Values Before Cleaning:")

for column in categorical_columns:
    print(column, ":", df[column].unique())


# Remove extra spaces and make capitalization consistent

for column in categorical_columns:
    df[column] = df[column].str.strip()
    df[column] = df[column].str.title()


print("\nUnique Values After Cleaning:")

for column in categorical_columns:
    print(column, ":", df[column].unique())
```

```python
# ==========================================================
# 4. HANDLING DATA TYPE ISSUES
# ==========================================================

print("\n--------------- DATA TYPE ISSUES ---------------")

print("Data Types Before Correction:")
print(df.dtypes)


numerical_columns = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity",
    "Exam_Score"
]


# Convert numerical columns to numeric datatype

for column in numerical_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")


print("\nData Types After Correction:")
print(df.dtypes)


# If conversion creates any missing values,
# replace them with the mean of that column

for column in numerical_columns:

    if df[column].isnull().sum() > 0:

        mean_value = df[column].mean()

        df[column] = df[column].fillna(mean_value)


```

```python
# ==========================================================
# 5. HANDLING OUTLIERS
# ==========================================================

print("\n--------------- OUTLIERS ---------------")

# Detect outliers using IQR method

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Lower Limit =", lower_limit)
    print("Upper Limit =", upper_limit)
    print("Number of Outliers =", len(outliers))

    if len(outliers) > 0:
        print("Outlier Values:")
        print(outliers[column].values)
```

```python
# ==========================================================
# REMOVE OUTLIERS FROM HOURS STUDIED
# ==========================================================

Q1 = df["Hours_Studied"].quantile(0.25)
Q3 = df["Hours_Studied"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df["Hours_Studied"] >= lower_limit) &
    (df["Hours_Studied"] <= upper_limit)
]
```

```python
# ==========================================================
# FINAL CLEANED DATASET
# ==========================================================

print("\n=============== FINAL CLEANED DATASET ===============")

display(df)

print("\nFinal Shape:")
print(df.shape)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

print("\nDATA PREPROCESSING COMPLETED SUCCESSFULLY")
```
