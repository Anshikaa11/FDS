```python
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("dataset.csv")


# -------------------- 1. HISTOGRAM --------------------

plt.hist(df["Exam_Score"], bins=10, edgecolor="black")
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.show()


# -------------------- 2. SCATTER PLOT --------------------

plt.scatter(df["Hours_Studied"], df["Exam_Score"])
plt.title("Hours Studied vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()


# -------------------- 3. BAR CHART --------------------

motivation = df["Motivation_Level"].value_counts()

plt.bar(motivation.index, motivation.values)
plt.title("Students by Motivation Level")
plt.xlabel("Motivation Level")
plt.ylabel("Number of Students")
plt.show()


# -------------------- 4. PIE CHART --------------------

school = df["School_Type"].value_counts()

plt.pie(school.values, labels=school.index, autopct="%1.1f%%")
plt.title("Distribution of School Type")
plt.show()
```
