```python
import pandas as pd
import numpy as np

# -------------------- READ DATASET --------------------
df = pd.read_csv("dataset.csv")
```

```python
def descriptive_stats(values):
    # Remove missing values if present
    data = [x for x in values if x is not None]

    n = len(data)
    if n == 0:
        return None

    # Sort for median, mode, min/max, and quantiles
    data.sort()

    # Mean
    total = 0
    for x in data:
        total += x
    mean = total / n

    # Median
    if n % 2 == 1:
        median = data[n // 2]
    else:
        median = (data[n // 2 - 1] + data[n // 2]) / 2

    # Mode(s)
    frequency = {}
    for x in data:
        if x in frequency:
            frequency[x] += 1
        else:
            frequency[x] = 1

    highest_frequency = max(frequency.values())
    modes = []

    for value in frequency:
        if frequency[value] == highest_frequency:
            modes.append(value)
 # Sample variance and sample standard deviation
    squared_difference_sum = 0

    for x in data:
        squared_difference_sum += (x - mean) ** 2

    sample_variance = squared_difference_sum / (n - 1)

    sample_std_dev = sample_variance ** 0.5

    # Minimum and maximum
    minimum = data[0]

    maximum = data[-1]
    
    range=maximum-minimum
    
    return {
        "count": n,
        "mean": mean,
        "median": median,
        "mode": modes,
        "mode_frequency": highest_frequency,
        "variance": sample_variance,
        "standard_deviation": sample_std_dev,
        "minimum": minimum,
        "maximum": maximum,
        "range": range}

result = descriptive_stats(df["Exam_Score"].tolist())

for name, values in result.items():
    print(name, ":", values)
```

```python
print("\n--------------- HOURS STUDIED OUTLIERS ---------------")

column = "Hours_Studied"

Q1 = df[column].quantile(0.25)

Q3 = df[column].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df[column] < lower_limit) |
    (df[column] > upper_limit)
]

print("Lower Limit =", lower_limit)

print("Upper Limit =", upper_limit)

print("Number of Outliers =", len(outliers))

if len(outliers) > 0:

    print("Outlier Values:")

    print(outliers[column].values)
```
